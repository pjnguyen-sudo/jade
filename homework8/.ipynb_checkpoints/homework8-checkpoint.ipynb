```python
df["latitude"].median()
```

```python
df["longitude"].median()
```

```python
df["dif_lon"] = df["longitude"] - df["longitude"].median() 
df["abs_lon"] = df["dif_lon"].abs() 
df
```

```python
df["dif_lat"] = df["latitude"] - df["latitude"].median() 
df["abs_lat"] = df["dif_lat"].abs() 
df
```

```python
# I have nine columns and 51 rows now.
```

```python
state_min_lat = df["abs_lat"].idxmin()
min_lat = df["abs_lat"].min()

print(f"State {state_min_lat} is closest to the median latitude, with a difference of {min_lat:.2f} degrees.")
```

```python
state_max_lat = df["abs_lat"].idxmax()
max_lat = df["abs_lat"].max()

print(f"State {state_max_lat} is farthest from the median latitude, with a difference of {max_lat:.2f} degrees.")
```

```python
state_min_lon = df["abs_lon"].idxmin()
min_lon = df["abs_lon"].min()

print(f"State {state_min_lon} is closest to the median longitude, with a difference of {min_lon:.2f} degrees.")
```

```python
state_max_lon = df["abs_lon"].idxmax()
max_lon = df["abs_lon"].max()

print(f"State {state_max_lon} is farthest from the median longitude, with a difference of {max_lon:.2f} degrees.")
```

```python
## Seaborn ##

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
car_crashes = sns.load_dataset("car_crashes")
df = pd.DataFrame(car_crashes)
df
```

```python
df.shape
```

```python
# The shape of the dataset is (51,8).
# There are 51 rows and eight columns.
```

```python
df.head(5)
```

```python
df.tail(5)
```

```python
# Total = total number of crashes
# Speeding = crashes caused by speeding
# Alcohol = crashes caused by alcohol
# Not_distracted = drivers not distracted
# No_previous = drivers with no previous crashes
# Ins_premium = state's average insurance premium
# Ins_losses = state's total insurance losses
# Abbrev = state abbreviation
```

```python
plt.figure(figsize=(2, 2))
abbrev_list = df["abbrev"].unique()
for abbrev in abbrev_list:
    df_subset = df[df["abbrev"] == abbrev]
    plt.scatter(df_subset["not_distracted"], df_subset["no_previous"], label=f"{abbrev}")

plt.title("Not Distracted vs. No Previous")
plt.xlabel("Not Distracted (%)")
plt.ylabel("No Previous (%)")
plt.legend()
plt.show()

plt.figure(figsize=(2, 2))
abbrev_list2 = df["abbrev"].unique()
for abbrev in abbrev_list:
    df_subset2 = df[df["abbrev"] == abbrev]
    plt.scatter(df_subset2["speeding"], df_subset2["alcohol"], label=f"{abbrev}")

plt.title("Alcohol vs. Speeding")
plt.xlabel("Alcohol (%)")
plt.ylabel("Speeding (%)")
plt.legend()
plt.show()

sns.histplot(
    data=car_crashes)

sns.boxplot(
    data=car_crashes)
```
