```python
# The different discovery methods are: Transit, Radial Velocity, Microlensing, Imaging, Transit Timing Variations, Eclipse Timing Variations, and Disk Kinematics.
# Of each, there are Transit (388), Radial Velocity (107), Microlensing (61), Imaging (5), Transit Timing Variations (3), Eclipse Timing Variations (2), Disk Kinematics (1). Transit has the most while Disk Kinematics has the least.
# Transit = when a planet passes between a star and its observer.
# Radial Velocity = the slight movements that affect a star's light spectrum when it orbits around a planet.
# Microlensing = aka gravitaional lensing, measures the distortion that occurs when stars and planets create a detectable degree of warping.
# Imaging = using coronagraphs to block starlight, making it easier to see fainter objects near a star.
# Transit Timing Variations = observing variations in the timing of a transit.
# Eclipse Timing Variations = measuring the timing of eclipses and looking for deviations from a constant period.
# Disk Kinematics = detecting the signatures produced by a planet's protoplanetary disk
```

```python
plt.figure(figsize=(8,6))
plt.xlabel("Planet Mass")
plt.ylabel("Equilibrium Temperature")
plt.title("Planet Mass vs. Equilibrium Temperature")

discovery_list = df["discoverymethod"].unique()
for discoverymethod in discovery_list:
    df_subset = df[df["discoverymethod"] == discoverymethod]
    plt.scatter(df_subset["pl_bmassj"], df_subset["st_teff"], label=f"{discoverymethod}")

plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.xlabel("Planet Mass")
plt.ylabel("Equilibrium Temperature")
plt.title("Planet Mass vs. Equilibrium Temperature")

discovery_list = df["discoverymethod"].unique()
for discoverymethod in discovery_list:
    df_subset = df[df["discoverymethod"] == discoverymethod]
    plt.scatter(np.log(df_subset["pl_bmassj"]), np.log(df_subset["st_teff"]), label=f"{discoverymethod}")

plt.legend()
plt.show()
```
